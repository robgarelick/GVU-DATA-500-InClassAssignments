# Exercise 2 (15 points)

In this exercise, we will practice documenting code the way Chapter 9 describes: with a descriptive name, type hints, and a complete Google-style docstring (summary, `Args`, `Returns`, `Raises`, and `Example`).

Consider the following undocumented, poorly named function:

```python
def f(df, n):
    df = df[df['price'] >= n]
    df['price'] = df['price'].astype(float)
    return df
```

Your task:

- **Rename the function** to something descriptive that follows the `verb_noun` convention (e.g., a function that filters rows should start with a verb like `filter_`).
- **Add type hints** to the parameters and the return value (the input and output are `pandas.DataFrame`; the threshold is a `float`).
- **Add a `ValueError`** that is raised if the threshold `n` is negative.
- **Write a complete Google-style docstring** including a one-line summary, `Args`, `Returns`, `Raises`, and a runnable `Example`.

Then apply your documented function to the following data, using a minimum price threshold of `20`.

```python
import pandas as pd

data = {'item': ['A', 'B', 'C', 'D'],
        'price': ['10', '25', '30', '15']}
df = pd.DataFrame(data)
```

The expected outcome is shown below.

| item | price |
| --- | --- |
| B | 25.0 |
| C | 30.0 |

````python
import pandas as pd


def filter_rows_by_min_price(df: pd.DataFrame, min_price: float) -> pd.DataFrame:
    """Keep only the rows whose price is greater than or equal to a minimum price.

    The price column is converted to float before filtering, so prices
    stored as text (for example '25') are handled correctly.

    Args:
        df (pd.DataFrame): The data containing at least 'price' column.
        min_price (float): The minimum price to filter rows by

    Returns:
        pd.DataFrame: A new DataFrame with only the rows where price is at
        least min_price, with price stored as float.

    Raises:
        ValueError: If min_price is negative.

    Example:
        >>> data = {'item': ['A', 'B', 'C', 'D'],
        ...         'price': ['10', '25', '30', '15']}
        >>> df = pd.DataFrame(data)
        >>> filter_rows_by_min_price(df, 20)
          item  price
        0    B   25.0
        1    C   30.0
    """
    if min_price < 0:
        raise ValueError("min_price must not be negative")

    result = df.copy()
    result['price'] = result['price'].astype(float)
    result = result[result['price'] >= min_price]
    return result.reset_index(drop=True)


# Apply the function to the data with a minimum price of 20
data = {'item': ['A', 'B', 'C', 'D'],
        'price': ['10', '25', '30', '15']}
df = pd.DataFrame(data)

filtered_df = filter_rows_by_min_price(df, 20)
print(filtered_df)
````
