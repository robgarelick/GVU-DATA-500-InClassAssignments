# Exercise 2 (10 points)

Data pre-processing with OOP. Create a Python class called `DataPreprocessor` that:

1. Takes a pandas DataFrame as input.
2. Provides methods to:
   - Handle missing values by filling them with the column mean.
   - Normalize numerical columns between 0 and 1.
3. Returns the cleaned DataFrame when called with `.transform()`

Consider `df` as DataFrame to test the class.

```
data = {
    "age": [25, np.nan, 35, 40, np.nan],
    "income": [50000, 60000, np.nan, 80000, 75000],
    "gender": ["M", "F", "M", "F", "M"]
}
df = pd.DataFrame(data)




````python
import numpy as np
import pandas as pd

class DataPreprocessor:
    """Initializes the DataPreporcessor with the given Dataframe"""
    def __init__(self, df):
        self.df = df.copy()   # work on a copy so the original stays untouched
        self.num_cols = self.df.select_dtypes(include='number').columns

    def fill_missing(self):
        """Fills the missing values of a given column the datafram with the colummn mean"""
        for col in self.num_cols:
            self.df[col] = self.df[col].fillna(self.df[col].mean())
        return self

    def normalize(self):
        """Normalize the numeric columns in the datafram to the rage [0, 1]"""
        for col in self.num_cols:
            col_min = self.df[col].min()
            col_max = self.df[col].max()
            if col_max != col_min:
                self.df[col] = (self.df[col] - col_min) / (col_max - col_min)
            else:
                self.df[col] = 0.0   # avoid dividing by zero on a constant column
        return self

    def transform(self):
        self.fill_missing()
        self.normalize()
        return self.df
    
    
data = {
    "age": [25, np.nan, 35, 40, np.nan],
    "income": [50000, 60000, np.nan, 80000, 75000],
    "gender": ["M", "F", "M", "F", "M"]
}
df = pd.DataFrame(data)

print("Original:")
display(df)

preprocessor = DataPreprocessor(df)
clean_df = preprocessor.transform()

print("Cleaned:")
display(clean_df.round(4))
````

### Exercise 4(a) (3 points)

Imagine you are working with a data set of customers that includes their purchase amounts in different categories. The goal is to apply transformations and extract insights using functional programming techniques. Create the below `pandas` data frame.

```
import pandas as pd

data = {

````python
import pandas as pd

data = {
    'Customer_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Groceries': [120, 150, 80, 90, 130],
    'Electronics': [200, 0, 100, 300, 150],
    'Clothing': [50, 60, 45, 70, 30]
}

df = pd.DataFrame(data)
df
````

### Exercise 4(b) (3 points)

Use the `apply()` function to calculate the total spending for each customer (sum of `Groceries`, `Electronics`, and `Clothing`).

````python
df['Total_Spending'] = df.apply(
    lambda row: row['Groceries'] + row['Electronics'] + row['Clothing'], axis=1
)
df
````

### Exercise 4(c) (3 points)

Use a `lambda` function to classify customers as `High`, `Medium`, or `Low` spenders based on their total spending:

- High: Total spending >= 400
- Medium: Total spending between 200 and 399
- Low: Total spending < 200

````python
df['Spender_Type'] = df['Total_Spending'].apply(
    lambda x: 'High' if x >= 400 else ('Medium' if x >= 200 else 'Low')
)
df[['Name', 'Total_Spending', 'Spender_Type']]
````

### Exercise 4(d) (3 points)

Calculate the percentage of spending in `Electronics` compared to the `Total_Spend` for each customer.

````python
df['Electronics_Pct'] = df.apply(
    lambda row: row['Electronics'] / row['Total_Spending'] * 100, axis=1
).round(2)
df[['Name', 'Electronics', 'Total_Spending', 'Electronics_Pct']]
````

### Exercise 4(e) (4 points)

Write a custom function that can be passed to `apply()` to give a 10% discount on electronics spending if a customer's total spend is above $400.

````python
def apply_discount(row):
    if row['Total_Spending'] > 400:
        return row['Electronics'] * 0.9
    return row['Electronics']

df['Electronics_Discounted'] = df.apply(apply_discount, axis=1)
df[['Name', 'Total_Spending', 'Electronics', 'Electronics_Discounted']]
````
