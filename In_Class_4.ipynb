# Filtering and Sorting Data

In this in-class assignment, we will practice some basic data filtering and sorting with `pandas`. We will use the `chipotle.csv` data file. This file contains information related chipotle sales.

# Exercise 1 (2 points) 

Load `pandas` and `numpy`

```python
import pandas as pd
import numpy as np
```

# Exercise 2 (2 points)

Create a data-frame called `chipotle` using the data from the `chipotle.csv`.

```python
chipotle = pd.read_csv("chipotle.csv")
chipotle.head(-1)
```

# Exercise 3 (8 points)

How many products cost more than $10.00?

```python
# Convert item_price to a number (safe to rerun: works on text or already-numeric values)
chipotle['item_price'] = pd.to_numeric(
    chipotle['item_price'].astype(str).str.replace('$', '', regex=False).str.strip()
)

# Price for a single unit of each item, item_price is multiplied by quantity, so we divide by quantity to get the unit price to be more accurate.
chipotle['unit_price'] = chipotle['item_price'] / chipotle['quantity']

# Distinct products whose single-unit price is over $10
over_10 = chipotle[chipotle['unit_price'] > 10.00]
over_10['item_name'].nunique()
```

# Exercise 4 (2 points)

Sort by the name of the item

```python
chipotle.sort_values(by='item_name', ascending=True)
```

# Exercise 5 (2 points)

What was the quantity of the most expensive item ordered?

```python
chipotle.sort_values(by='item_price', ascending=False).head(1)
```

# Exercise 6 (2 points)

How many times was a Veggie Salad Bowl ordered?

```python
veggie_salad_order_count = chipotle[chipotle['item_name'] == 'Veggie Salad Bowl']
len(veggie_salad_order_count)
```

# Exercise 7 (3 points)

How many times did someone order more than one Canned Soda?

```python
canned_soda_filtered_orders = chipotle[(chipotle['item_name'] == 'Canned Soda') & (chipotle['quantity'] > 1)]
len(canned_soda_filtered_orders)
```
